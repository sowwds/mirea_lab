## 2.2. Решите уравнение и изобразите множество решений на комплексной плоскости: $z^8 = 1 + i$

## 2.3. Вычислите и изобразите на комплексной плоскости:
а) $\sqrt[8]{1}$ &nbsp;&nbsp; б) $\sqrt[4]{\dfrac{-18}{1 + i\sqrt{3}}}$ &nbsp;&nbsp; в) $\sqrt[3]{\dfrac{1-5i}{1+i} -51 + \dfrac{2i}{2-i} + 11}$ &nbsp;&nbsp; г) $\sqrt[9]{1-i}$

```python
import sympy as sp
import matplotlib.pyplot as plt

I = sp.I
pi = sp.pi

def nth_roots(z, n):
    """Все n-тые корни комплексного числа z (список sympy-комплексных)."""
    r  = sp.Abs(z)
    φ  = sp.arg(z)
    ρ  = r**(sp.Rational(1, n))
    return [sp.simplify(ρ*(sp.cos((φ + 2*pi*k)/n) + I*sp.sin((φ + 2*pi*k)/n)))
            for k in range(n)]

def print_roots(roots, label):
    print(f"\n=== {label} ===")
    for k, zk in enumerate(roots, 1):
        print(f"z{k} = {sp.N(zk)}")

def plot_roots(ax, roots, title):
    xs = [sp.re(z).evalf() for z in roots]
    ys = [sp.im(z).evalf() for z in roots]
    ax.scatter(xs, ys, s=30)
    ax.axhline(0, color='0.3', lw=0.8)
    ax.axvline(0, color='0.3', lw=0.8)
    ax.set_aspect('equal', adjustable='box')
    ax.grid(True, ls=':')
    ax.set_title(title)
    for k, (x, y) in enumerate(zip(xs, ys), 1):
        ax.text(x+0.05, y, f"{k}", fontsize=8)

za = 1
roots_a = nth_roots(za, 8)
print_roots(roots_a, "a) 8-е корни 1")

zb = -sp.Rational(18,1)/(1 + I*sp.sqrt(3))
zb = sp.simplify(zb)
roots_b = nth_roots(zb, 4)
print(f"\nУпрощённое подкоренное b): {zb}  = {sp.Abs(zb)}·exp(i·{sp.arg(zb)})")
print_roots(roots_b, "b) 4-е корни числа  -18/(1+i√3)")

zc = (1-5*I)/(1+I) - 5*(1+2*I)/(2-I) + 11
zc = sp.simplify(zc)
roots_c = nth_roots(zc, 3)
print(f"\nУпрощённое подкоренное c): {zc}  = {sp.Abs(zc)}·exp(i·{sp.arg(zc)})")
print_roots(roots_c, "c) 3-и корни")

zd = 1 - I
roots_d = nth_roots(zd, 9)
print_roots(roots_d, "d) 9-е корни (1 - i)")

fig, axs = plt.subplots(2, 2, figsize=(8, 8))

plot_roots(axs[0,0], roots_a, "a) 8-е корни 1")
plot_roots(axs[0,1], roots_b, "b) 4-е корни  −18/(1+i√3)")
plot_roots(axs[1,0], roots_c, "c) 3-и корни выражения")
plot_roots(axs[1,1], roots_d, "d) 9-е корни (1−i)")

plt.tight_layout()
plt.show()
```

## 2.4. Изобразите множества точек:
а) $|z-i| + |z+i| < 4$ &nbsp;&nbsp; б) $\operatorname{Re}(z(1-i)) < \sqrt{2}$ &nbsp;&nbsp; в) $|z+1| = |z+i| = |z-i|$  
г) $0 < \arg\frac{i-z}{z+i} < \frac{\pi}{2}$ &nbsp;&nbsp; д) $-\frac{2\pi}{3} \le \arg\frac{2-i z}{z+3i} \le -\frac{5\pi}{6}$ &nbsp;&nbsp; е) $-\frac{2\pi}{3} \le \arg\frac{3-z}{z+i} \le -\frac{\pi}{3}$

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import Ellipse, Arc
import warnings
warnings.filterwarnings('ignore')

# Настройка стиля
plt.rcParams['figure.figsize'] = [15, 12]
plt.rcParams['font.size'] = 12

def plot_complex_sets():
    """
    Визуализация множеств точек на комплексной плоскости
    """
    # Создаем сетку точек на комплексной плоскости
    x = np.linspace(-3, 3, 500)
    y = np.linspace(-3, 3, 500)
    X, Y = np.meshgrid(x, y)
    Z = X + 1j * Y

    # Создаем фигуру с 6 подграфиками
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle('Множества точек на комплексной плоскости', fontsize=16, fontweight='bold')

    # a) |z - i| + |z + i| < 4 (Эллипс)
    print("a) |z - i| + |z + i| < 4 - Эллипс с фокусами в i и -i")
    condition_a = np.abs(Z - 1j) + np.abs(Z + 1j) < 4
    im_a = axes[0, 0].imshow(condition_a, extent=[-3, 3, -3, 3], origin='lower',
                            cmap='Blues', alpha=0.7)
    axes[0, 0].plot(0, 1, 'ro', markersize=8, label='z = i')  # Фокус 1
    axes[0, 0].plot(0, -1, 'ro', markersize=8, label='z = -i')  # Фокус 2
    axes[0, 0].set_title('a) $|z - i| + |z + i| < 4$\n(Эллипс)')
    axes[0, 0].set_xlabel('Re(z)')
    axes[0, 0].set_ylabel('Im(z)')
    axes[0, 0].grid(True, alpha=0.3)
    axes[0, 0].legend()

    # b) Re(z(1 - i)) < √2
    print("b) Re(z(1 - i)) < √2 - Полуплоскость")
    # z(1 - i) = (x + iy)(1 - i) = (x + y) + i(y - x)
    # Re(z(1 - i)) = x + y
    condition_b = X + Y < np.sqrt(2)
    im_b = axes[0, 1].imshow(condition_b, extent=[-3, 3, -3, 3], origin='lower',
                            cmap='Greens', alpha=0.7)
    # Рисуем граничную прямую x + y = √2
    x_line = np.linspace(-3, 3, 100)
    y_line = np.sqrt(2) - x_line
    axes[0, 1].plot(x_line, y_line, 'r--', linewidth=2, label='$x + y = \sqrt{2}$')
    axes[0, 1].set_title('b) $Re(z(1 - i)) < \sqrt{2}$\n(Полуплоскость)')
    axes[0, 1].set_xlabel('Re(z)')
    axes[0, 1].set_ylabel('Im(z)')
    axes[0, 1].grid(True, alpha=0.3)
    axes[0, 1].legend()

    # c) |z + 1| = |z + i| = |z - i|
    print("c) |z + 1| = |z + i| = |z - i| - Точка пересечения окружностей")
    # Это система уравнений, решаем аналитически
    # |z + 1| = |z + i| => (x+1)² + y² = x² + (y+1)² => x = y
    # |z + 1| = |z - i| => (x+1)² + y² = x² + (y-1)² => x = -y
    # Решение: x = y = 0 => z = 0

    condition_c = (np.abs(Z + 1) - np.abs(Z + 1j) < 0.1) & (np.abs(Z + 1) - np.abs(Z - 1j) < 0.1)
    im_c = axes[0, 2].imshow(condition_c, extent=[-3, 3, -3, 3], origin='lower',
                            cmap='Reds', alpha=0.7)
    axes[0, 2].plot(0, 0, 'ro', markersize=10, label='z = 0')
    axes[0, 2].plot(-1, 0, 'g^', markersize=8, label='z = -1')  # Центр 1
    axes[0, 2].plot(0, -1, 'g^', markersize=8, label='z = -i')  # Центр 2
    axes[0, 2].plot(0, 1, 'g^', markersize=8, label='z = i')    # Центр 3
    axes[0, 2].set_title('c) $|z + 1| = |z + i| = |z - i|$\n(Точка z = 0)')
    axes[0, 2].set_xlabel('Re(z)')
    axes[0, 2].set_ylabel('Im(z)')
    axes[0, 2].grid(True, alpha=0.3)
    axes[0, 2].legend()

    # d) 0 < arg((i - z)/(z + i)) < π/2
    print("d) 0 < arg((i - z)/(z + i)) < π/2 - Угловой сектор")
    # (i - z)/(z + i) = (i - (x + iy))/(x + i(y + 1)) = (-x + i(1 - y))/(x + i(y + 1))
    w_d = (1j - Z) / (Z + 1j)
    arg_d = np.angle(w_d)
    condition_d = (arg_d > 0) & (arg_d < np.pi/2)
    im_d = axes[1, 0].imshow(condition_d, extent=[-3, 3, -3, 3], origin='lower',
                            cmap='Purples', alpha=0.7)
    axes[1, 0].plot(0, 1, 'ro', markersize=8, label='z = i')
    axes[1, 0].plot(0, -1, 'ro', markersize=8, label='z = -i')
    axes[1, 0].set_title('d) $0 < arg\\frac{i - z}{z + i} < \\frac{\\pi}{2}$\n(Угловой сектор)')
    axes[1, 0].set_xlabel('Re(z)')
    axes[1, 0].set_ylabel('Im(z)')
    axes[1, 0].grid(True, alpha=0.3)
    axes[1, 0].legend()

    # e) -2π/3 ≤ arg((2 - iz)/(z + 3i)) ≤ -5π/6
    print("e) -2π/3 ≤ arg((2 - iz)/(z + 3i)) ≤ -5π/6 - Угловой сектор")
    # 2 - iz = 2 - i(x + iy) = (2 + y) - ix
    # z + 3i = x + i(y + 3)
    w_e = (2 - 1j*Z) / (Z + 3j)
    arg_e = np.angle(w_e)
    condition_e = (arg_e >= -2*np.pi/3) & (arg_e <= -5*np.pi/6)
    im_e = axes[1, 1].imshow(condition_e, extent=[-3, 3, -3, 3], origin='lower',
                            cmap='Oranges', alpha=0.7)
    axes[1, 1].set_title('e) $-\\frac{2\\pi}{3} \\leq arg\\frac{2 - iz}{z + 3i} \\leq -\\frac{5\\pi}{6}$\n(Угловой сектор)')
    axes[1, 1].set_xlabel('Re(z)')
    axes[1, 1].set_ylabel('Im(z)')
    axes[1, 1].grid(True, alpha=0.3)

    # f) -2π/3 ≤ arg((3 - z)/(z + i)) ≤ -π/3
    print("f) -2π/3 ≤ arg((3 - z)/(z + i)) ≤ -π/3 - Угловой сектор")
    w_f = (3 - Z) / (Z + 1j)
    arg_f = np.angle(w_f)
    condition_f = (arg_f >= -2*np.pi/3) & (arg_f <= -np.pi/3)
    im_f = axes[1, 2].imshow(condition_f, extent=[-3, 3, -3, 3], origin='lower',
                            cmap='RdPu', alpha=0.7)
    axes[1, 2].plot(3, 0, 'ro', markersize=8, label='z = 3')
    axes[1, 2].plot(0, -1, 'ro', markersize=8, label='z = -i')
    axes[1, 2].set_title('f) $-\\frac{2\\pi}{3} \\leq arg\\frac{3 - z}{z + i} \\leq -\\frac{\\pi}{3}$\n(Угловой сектор)')
    axes[1, 2].set_xlabel('Re(z)')
    axes[1, 2].set_ylabel('Im(z)')
    axes[1, 2].grid(True, alpha=0.3)
    axes[1, 2].legend()

    plt.tight_layout()
    plt.show()

    # Вывод аналитических решений
    print_analytical_solutions()

def print_analytical_solutions():
    """
    Вывод аналитических решений
    """
    print("\n" + "="*80)
    print("АНАЛИТИЧЕСКИЕ РЕШЕНИЯ:")
    print("="*80)

    print("a) |z - i| + |z + i| < 4")
    print("   Эллипс с фокусами в точках i и -i")
    print("   Большая полуось: a = 2")
    print("   Малая полуось: b = √3")
    print("   Уравнение: x²/3 + y²/4 = 1")
    print()

    print("b) Re(z(1 - i)) < √2")
    print("   z(1 - i) = (x + iy)(1 - i) = (x + y) + i(y - x)")
    print("   Re(z(1 - i)) = x + y")
    print("   Неравенство: x + y < √2")
    print("   Полуплоскость ниже прямой x + y = √2")
    print()

    print("c) |z + 1| = |z + i| = |z - i|")
    print("   |z + 1| = |z + i| => (x+1)² + y² = x² + (y+1)² => x = y")
    print("   |z + 1| = |z - i| => (x+1)² + y² = x² + (y-1)² => x = -y")
    print("   Решение: x = 0, y = 0 => z = 0")
    print()

    print("d) 0 < arg((i - z)/(z + i)) < π/2")
    print("   Дробно-линейное преобразование")
    print("   Точки: z₁ = i, z₂ = -i")
    print("   Угловой сектор между лучами из точки z = 0")
    print()

    print("e) -2π/3 ≤ arg((2 - iz)/(z + 3i)) ≤ -5π/6")
    print("   Преобразование: 2 - iz = (2 + y) - ix")
    print("   z + 3i = x + i(y + 3)")
    print("   Угловой сектор на комплексной плоскости")
    print()

    print("f) -2π/3 ≤ arg((3 - z)/(z + i)) ≤ -π/3")
    print("   Точки: z₁ = 3, z₂ = -i")
    print("   Угловой сектор между указанными углами")
    print("="*80)

# Запускаем визуализацию
plot_complex_sets()

# Дополнительная детализированная визуализация для некоторых случаев
def detailed_visualization():
    """
    Детализированная визуализация для сложных случаев
    """
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

    # Детализация для пункта a (эллипс)
    x = np.linspace(-2, 2, 500)
    y = np.linspace(-2, 2, 500)
    X, Y = np.meshgrid(x, y)
    Z = X + 1j * Y

    # a) Эллипс с аналитическим решением
    condition_a = np.abs(Z - 1j) + np.abs(Z + 1j) < 4
    im1 = ax1.imshow(condition_a, extent=[-2, 2, -2, 2], origin='lower', cmap='Blues')

    # Аналитическое решение эллипса: x²/b² + y²/a² = 1, где a=2, b=√3
    theta = np.linspace(0, 2*np.pi, 100)
    x_ellipse = np.sqrt(3) * np.cos(theta)
    y_ellipse = 2 * np.sin(theta)
    ax1.plot(x_ellipse, y_ellipse, 'r--', linewidth=2, label='Аналитическое решение')

    ax1.plot(0, 1, 'ro', markersize=8, label='Фокус: i')
    ax1.plot(0, -1, 'ro', markersize=8, label='Фокус: -i')
    ax1.set_title('a) Эллипс: $|z - i| + |z + i| < 4$\n$x^2/3 + y^2/4 = 1$')
    ax1.set_xlabel('Re(z)')
    ax1.set_ylabel('Im(z)')
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    ax1.axis('equal')

    # Детализация для пункта b (полуплоскость)
    condition_b = X + Y < np.sqrt(2)
    im2 = ax2.imshow(condition_b, extent=[-2, 2, -2, 2], origin='lower', cmap='Greens')

    # Граничная прямая
    x_line = np.linspace(-2, 2, 100)
    y_line = np.sqrt(2) - x_line
    ax2.plot(x_line, y_line, 'r-', linewidth=2, label='$x + y = \sqrt{2}$')

    ax2.set_title('b) $Re(z(1 - i)) < \sqrt{2}$\n$z = x + iy$')
    ax2.set_xlabel('Re(z)')
    ax2.set_ylabel('Im(z)')
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    ax2.axis('equal')

    plt.tight_layout()
    plt.show()

detailed_visualization()
```

## 2.5. Покажите, что уравнение $A|z|^2 + Bz + \overline{B}z + C = 0$ (A>0, AC < |B|^2) — окружность

```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np

A = 2
B = 3 - 4*sp.I
C = 1

center = -B / A
radius = sp.sqrt(abs(B)**2 - A*C) / A

print("=== Уравнение окружности ===")
print(f"A|z|² + B̅z + Bz̅ + C = 0")
print(f"При A = {A}, B = {B}, C = {C}")
print()
print(f"Центр: z₀ = -B/A = {center}")
print(f"Радиус: R = √(|B|² - AC)/A = {radius.simplify()}")
print()

# Проверим, что окружность существует
if abs(B)**2 - A*C <= 0:
    print("AC ≥ |B|² — окружность не существует (радиус мнимый).")
else:
    print("AC < |B|² — окружность существует.")

x0 = float(sp.re(center))
y0 = float(sp.im(center))
r = float(radius)

theta = np.linspace(0, 2*np.pi, 300)
x = x0 + r * np.cos(theta)
y = y0 + r * np.sin(theta)

plt.figure(figsize=(5,5))
plt.plot(x, y, 'b', label=f"Окружность (R={r:.2f})")
plt.scatter(x0, y0, color='red', label=f"Центр ({x0:.2f}, {y0:.2f})")

plt.axhline(0, color='black', lw=0.8)
plt.axvline(0, color='black', lw=0.8)
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True, ls=':')
plt.title("Окружность:  A|z|² + B̅z + Bz̅ + C = 0")
plt.legend()
plt.show()
```

## 2.6. Найти все корни многочлена $p(x) = x^{11} + x^{10} + x^8 - 2x^7 - x^5 + x^4 + 2x^2 + 2x + 2$

```python
import sympy as sp
import numpy as np

x = sp.Symbol('x')

p = x**11 + x**10 + x**8 - 2*x**7 - x**5 + x**4 + 2*x**2 + 2*x + 2

roots = sp.nroots(p)

print("=== Корни многочлена p(x) ===")
for i, z in enumerate(roots, start=1):
    print(f"z{i} = {z}")

sum_mod = sum(abs(complex(z)) for z in roots)
print(f"\n1) Сумма модулей |z₁| + ... + |z₁₁| = {sum_mod:.5f}")

roots_complex = [complex(z) for z in roots]
z_max = max(roots_complex, key=lambda z: z.real)
print(f"2) Максимальная действительная часть = {z_max.real:.5f}")
print(f"   Этот корень: z = {z_max}")

import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
plt.scatter([z.real for z in roots_complex],
            [z.imag for z in roots_complex],
            color='blue')

plt.axhline(0, color='black', lw=0.8)
plt.axvline(0, color='black', lw=0.8)
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True, ls=':')
plt.title("Корни многочлена p(x) на комплексной плоскости")
plt.xlabel("Re(z)")
plt.ylabel("Im(z)")
plt.show()
```

## 2.7. Обобщённые биномиальные коэффициенты

```python
from mpmath import gamma, arg, mpc, mp
import numpy as np
import matplotlib.pyplot as plt

mp.dps = 20
comp_z = mpc(1,1)
comp_w = mpc(1,-1)
binom_gen = gamma(comp_z + 1) / (gamma(comp_w + 1) * gamma(comp_z - comp_w + 1))
print("a)", complex(binom_gen))
print("b)", arg(binom_gen))

grid_x = np.linspace(-2,2,150)
grid_y = np.linspace(-2,2,150)
mesh_x, mesh_y = np.meshgrid(grid_x, grid_y)
arg_vals = np.zeros_like(mesh_x)
for idx_i in range(mesh_x.shape[0]):
    for idx_j in range(mesh_x.shape[1]):
        point_z = mpc(mesh_x[idx_i,idx_j], mesh_y[idx_i,idx_j])
        binom_1 = gamma(point_z + 1) / (gamma(2) * gamma(point_z))
        arg_vals[idx_i,idx_j] = arg(binom_1)

plt.contour(mesh_x, mesh_y, arg_vals, levels=15)
plt.colorbar()
plt.title('c) Контуры Arg')
plt.xlabel('Re')
plt.ylabel('Im')
plt.show()
```

## 2.9. Множества Жюлиа

```python
import numpy as np
import matplotlib.pyplot as plt
from cmath import *

im_width, im_height = 800, 800
xmin, xmax = -2, 2
ymin, ymax = -2, 2
max_iter = 100

def julia_set(f, title):
    x = np.linspace(xmin, xmax, im_width)
    y = np.linspace(ymin, ymax, im_height)
    julia = np.zeros((im_width, im_height))

    for ix in range(im_width):
        for iy in range(im_height):
            z = complex(x[ix], y[iy])
            n = 0
            while abs(z) < 10 and n < max_iter:
                z = f(z)
                n += 1
            julia[ix, iy] = n / max_iter

    plt.figure(figsize=(7,7))
    plt.imshow(julia.T, extent=[xmin, xmax, ymin, ymax],
               origin='lower', cmap='magma')
    plt.title(title)
    plt.axis('off')
    plt.show()

def f1(z):
    return (1j * z**5 - 1) / (z**5 - 1j)

julia_set(f1, r"$f(z) = \frac{i z^5 - 1}{z^5 - i}$")

def f2(z):
    return (z**2 + 1j) / (1 - z**2)

julia_set(f2, r"$f(z) = \frac{z^2 + i}{1 - z^2}$")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors

def julia_set(width, height, c, x_min, x_max, y_min, y_max, max_iter=100):
    """
    Генерация множества Жюлиа для функции f(z) = z^2 + c
    """
    x = np.linspace(x_min, x_max, width)
    y = np.linspace(y_min, y_max, height)
    X, Y = np.meshgrid(x, y)

    Z = X + 1j * Y
    iterations = np.zeros(Z.shape, dtype=int)
    for i in range(max_iter):
        mask = np.abs(Z) <= 2
        Z[mask] = Z[mask]**2 + c
        iterations[mask] = i

    return iterations

c1 = -0.1 + 0.657j
c2 = 0.28 + 0.009j

width, height = 800, 800
x_min, x_max = -2, 2
y_min, y_max = -2, 2
max_iter = 100

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

julia1 = julia_set(width, height, c1, x_min, x_max, y_min, y_max, max_iter)
im1 = ax1.imshow(julia1, extent=[x_min, x_max, y_min, y_max],
                cmap='hot', origin='lower')
ax1.set_title(f'Множество Жюлиа для c = {c1:.3f}')
ax1.set_xlabel('Re(z)')
ax1.set_ylabel('Im(z)')
plt.colorbar(im1, ax=ax1, label='Количество итераций')

julia2 = julia_set(width, height, c2, x_min, x_max, y_min, y_max, max_iter)
im2 = ax2.imshow(julia2, extent=[x_min, x_max, y_min, y_max],
                cmap='viridis', origin='lower')
ax2.set_title(f'Множество Жюлиа для c = {c2:.3f}')
ax2.set_xlabel('Re(z)')
ax2.set_ylabel('Im(z)')
plt.colorbar(im2, ax=ax2, label='Количество итераций')

plt.tight_layout()
plt.show()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

im1 = ax1.imshow(julia1, extent=[x_min, x_max, y_min, y_max],
                cmap='plasma', origin='lower')
ax1.set_title(f'Множество Жюлиа для c = {c1:.3f} (альтернативная схема)')
ax1.set_xlabel('Re(z)')
ax1.set_ylabel('Im(z)')
plt.colorbar(im1, ax=ax1, label='Количество итераций')

im2 = ax2.imshow(julia2, extent=[x_min, x_max, y_min, y_max],
                cmap='inferno', origin='lower')
ax2.set_title(f'Множество Жюлиа для c = {c2:.3f} (альтернативная схема)')
ax2.set_xlabel('Re(z)')
ax2.set_ylabel('Im(z)')
plt.colorbar(im2, ax=ax2, label='Количество итераций')

plt.tight_layout()
plt.show()

print("Параметры множеств Жюлиа:")
print(f"c1 = {c1}")
print(f"c2 = {c2}")
print(f"Диапазон по x: [{x_min}, {x_max}]")
print(f"Диапазон по y: [{y_min}, {y_max}]")
print(f"Максимальное количество итераций: {max_iter}")
```

## 2.10. Стереографическая проекция на сферу Римана

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import warnings
warnings.filterwarnings('ignore')

def complex_to_sphere(z):
    """
    Стереографическая проекция: отображает комплексное число на сферу Римана
    """
    x = np.real(z)
    y = np.imag(z)
    r_sq = x**2 + y**2
    X = 2*x / (1 + r_sq)
    Y = 2*y / (1 + r_sq)
    Z = (r_sq - 1) / (1 + r_sq)
    return X, Y, Z

def plot_riemann_sphere_with_curves():
    """
    Визуализация кривых на сфере Римана в 3D
    """
    fig = plt.figure(figsize=(16, 12))

    u = np.linspace(0, 2 * np.pi, 50)
    v = np.linspace(0, np.pi, 25)
    x_sphere = np.outer(np.cos(u), np.sin(v))
    y_sphere = np.outer(np.sin(u), np.sin(v))
    z_sphere = np.outer(np.ones(np.size(u)), np.cos(v))

    ax1 = fig.add_subplot(221, projection='3d')

    ax1.plot_surface(x_sphere, y_sphere, z_sphere, color='lightblue',
                    alpha=0.3, rstride=1, cstride=1)

    print("а) Прямая линия: z(t) = t + it")
    t_line = np.linspace(-3, 3, 100)
    z_line = t_line + 1j * t_line
    X_line, Y_line, Z_line = complex_to_sphere(z_line)
    ax1.plot(X_line, Y_line, Z_line, 'b-', linewidth=3, label='Прямая')

    ax1.set_title('а) Прямая на сфере Римана\n$z(t) = t + it$')
    ax1.set_xlabel('X')
    ax1.set_ylabel('Y')
    ax1.set_zlabel('Z')
    ax1.legend()

    ax2 = fig.add_subplot(222, projection='3d')
    ax2.plot_surface(x_sphere, y_sphere, z_sphere, color='lightblue',
                    alpha=0.3, rstride=1, cstride=1)

    print("б) Окружность: z(t) = e^(it)")
    t_circle = np.linspace(0, 2*np.pi, 100)
    z_circle = np.exp(1j * t_circle)
    X_circle, Y_circle, Z_circle = complex_to_sphere(z_circle)
    ax2.plot(X_circle, Y_circle, Z_circle, 'r-', linewidth=3, label='Окружность')

    ax2.set_title('б) Окружность на сфере Римана\n$z(t) = e^{it}$')
    ax2.set_xlabel('X')
    ax2.set_ylabel('Y')
    ax2.set_zlabel('Z')
    ax2.legend()

    ax3 = fig.add_subplot(223, projection='3d')
    ax3.plot_surface(x_sphere, y_sphere, z_sphere, color='lightblue',
                    alpha=0.3, rstride=1, cstride=1)

    print("в) Парабола: z(t) = t + it²")
    t_parabola = np.linspace(-2, 2, 100)
    z_parabola = t_parabola + 1j * t_parabola**2
    X_parabola, Y_parabola, Z_parabola = complex_to_sphere(z_parabola)
    ax3.plot(X_parabola, Y_parabola, Z_parabola, 'g-', linewidth=3, label='Парабола')

    ax3.set_title('в) Парабола на сфере Римана\n$z(t) = t + it^2$')
    ax3.set_xlabel('X')
    ax3.set_ylabel('Y')
    ax3.set_zlabel('Z')
    ax3.legend()

    ax4 = fig.add_subplot(224, projection='3d')
    ax4.plot_surface(x_sphere, y_sphere, z_sphere, color='lightblue',
                    alpha=0.3, rstride=1, cstride=1)

    print("г) Спираль (перо птицы): z(t) = te^(it)")
    t_spiral = np.linspace(0.1, 4*np.pi, 200)
    z_spiral = t_spiral * np.exp(1j * t_spiral)
    X_spiral, Y_spiral, Z_spiral = complex_to_sphere(z_spiral)
    ax4.plot(X_spiral, Y_spiral, Z_spiral, 'purple', linewidth=3, label='Спираль')

    ax4.set_title('г) Спираль на сфере Римана\n$z(t) = te^{it}$')
    ax4.set_xlabel('X')
    ax4.set_ylabel('Y')
    ax4.set_zlabel('Z')
    ax4.legend()

    plt.tight_layout()
    plt.show()

    fig_all = plt.figure(figsize=(12, 10))
    ax_all = fig_all.add_subplot(111, projection='3d')


    ax_all.plot_surface(x_sphere, y_sphere, z_sphere, color='lightblue',
                       alpha=0.2, rstride=1, cstride=1)


    ax_all.plot(X_line, Y_line, Z_line, 'b-', linewidth=2, label='Прямая: $z(t) = t + it$')
    ax_all.plot(X_circle, Y_circle, Z_circle, 'r-', linewidth=2, label='Окружность: $z(t) = e^{it}$')
    ax_all.plot(X_parabola, Y_parabola, Z_parabola, 'g-', linewidth=2, label='Парабола: $z(t) = t + it^2$')
    ax_all.plot(X_spiral, Y_spiral, Z_spiral, 'purple', linewidth=2, label='Спираль: $z(t) = te^{it}$')

    ax_all.set_title('Все кривые на сфере Римана', fontsize=14, fontweight='bold')
    ax_all.set_xlabel('X')
    ax_all.set_ylabel('Y')
    ax_all.set_zlabel('Z')
    ax_all.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

    plt.tight_layout()
    plt.show()

plot_riemann_sphere_with_curves()

print("\n" + "="*70)
print("СТЕРЕОГРАФИЧЕСКАЯ ПРОЕКЦИЯ НА СФЕРУ РИМАНА")
print("="*70)
print("Формулы проекции:")
print("Для комплексного числа z = x + iy:")
print("X = 2x / (1 + x² + y²)")
print("Y = 2y / (1 + x² + y²)")
print("Z = (x² + y² - 1) / (1 + x² + y²)")
print("\nСеверный полюс (0,0,1) соответствует бесконечности")
print("Южный полюс (0,0,-1) соответствует нулю")
print("Экватор соответствует единичной окружности |z| = 1")
print("="*70)


```
