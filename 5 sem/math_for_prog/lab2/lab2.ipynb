## 2.1. Представьте в алгебраической, тригонометрической и показательной формах следующие комплексные выражения:

а) $\dfrac{(5 + i)(7 - 6i)}{3 + i}$  &nbsp;&nbsp;&nbsp;&nbsp; б) $\dfrac{(1 + i)^5}{(1 - i)^3}$  &nbsp;&nbsp;&nbsp;&nbsp; в) $\dfrac{(-1 + i\sqrt{3})^{15}}{(1 - i)^{20}} + \dfrac{(-1 - i\sqrt{3})^{15}}{(1 + i)^{20}}$.

```python
import numpy as np
import cmath
import matplotlib.pyplot as plt
from scipy.special import gamma
plt.rcParams['figure.dpi'] = 100

print("а)")
z1 = (5 + 1j)*(7 - 6j) / (3 + 1j)
print(f"Алгебраическая: {z1}")
print(f"Тригонометрическая: {abs(z1):.6f} (cos({np.angle(z1):.6f}) + i sin({np.angle(z1):.6f}))")
print(f"Показательная: {abs(z1):.6f} * exp(i * {np.angle(z1):.6f})")

print("\nб)")
z2 = (1 + 1j)**5 / (1 - 1j)**3
print(f"Алгебраическая: {z2}")
print(f"Тригонометрическая: {abs(z2):.6f} (cos({np.angle(z2):.6f}) + i sin({np.angle(z2):.6f}))")
print(f"Показательная: {abs(z2):.6f} * exp(i * {np.angle(z2):.6f})")

print("\nв)")
w1 = (-1 + 1j*np.sqrt(3))**15 / (1 - 1j)**20
w2 = (-1 - 1j*np.sqrt(3))**15 / (1 + 1j)**20
z3 = w1 + w2
print(f"Результат: {z3}")
```

## 2.2. Решите уравнение и изобразите множество решений на комплексной плоскости: $z^8 = 1 + i$

```python
r = np.sqrt(2)**(1/8)
roots = [r * cmath.exp(1j*(np.pi/4 + 2*np.pi*k)/8) for k in range(8)]
for i, root in enumerate(roots):
    print(f"z_{i+1} = {root}")

plt.figure(figsize=(6,6))
plt.scatter([z.real for z in roots], [z.imag for z in roots], color='red')
circle = plt.Circle((0,0), (2)**0.5, fill=False, color='blue', ls='--', label='|z| = √2^{1/8}')
plt.gca().add_patch(circle)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.grid(True)
plt.axis('equal')
plt.title("Корни уравнения z^8 = 1 + i")
plt.legend()
plt.show()
```

## 2.4. Изобразите множества точек:
а) $|z-i| + |z+i| < 4$ &nbsp;&nbsp; б) $\operatorname{Re}(z(1-i)) < \sqrt{2}$ &nbsp;&nbsp; в) $|z+1| = |z+i| = |z-i|$  
г) $0 < \arg\frac{i-z}{z+i} < \frac{\pi}{2}$ &nbsp;&nbsp; д) $-\frac{2\pi}{3} \le \arg\frac{2-i z}{z+3i} \le -\frac{5\pi}{6}$ &nbsp;&nbsp; е) $-\frac{2\pi}{3} \le \arg\frac{3-z}{z+i} \le -\frac{\pi}{3}$

```python
x = np.linspace(-4, 4, 500)
y = np.linspace(-4, 4, 500)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

plt.figure(figsize=(18,10))

plt.subplot(2,3,1)
plt.contourf(X,Y,abs(Z-1j)+abs(Z+1j)<4, levels=[0,1], colors='lightblue')
plt.title("а) |z−i| + |z+i| < 4")
plt.axis('equal')

plt.subplot(2,3,2)
plt.contourf(X,Y,np.real(Z*(1-1j))<np.sqrt(2), levels=[0,1], colors='lightgreen')
plt.title("б) Re(z(1−i)) < √2")
plt.axis('equal')

plt.subplot(2,3,3)
cond = np.isclose(abs(Z+1), abs(Z+1j)) & np.isclose(abs(Z+1j), abs(Z-1j))
plt.scatter(X[cond], Y[cond], s=1, color='red')
plt.title("в) |z+1|=|z+i|=|z−i|")
plt.axis('equal')

plt.subplot(2,3,4)
arg1 = np.angle((1j - Z)/(Z + 1j))
plt.contourf(X,Y,(arg1>0)&(arg1<np.pi/2), colors='gold')
plt.title("г) 0 < arg((i−z)/(z+i)) < π/2")
plt.axis('equal')

plt.subplot(2,3,5)
arg2 = np.angle((2 - 1j*Z)/(Z + 3j))
plt.contourf(X,Y,(arg2>=-2*np.pi/3)&(arg2<=-5*np.pi/6), colors='purple')
plt.title("д)")
plt.axis('equal')

plt.subplot(2,3,6)
arg3 = np.angle((3-Z)/(Z + 1j))
plt.contourf(X,Y,(arg3>=-2*np.pi/3)&(arg3<=-np.pi/3), colors='orange')
plt.title("е)")
plt.axis('equal')

plt.tight_layout()
plt.show()
```

## 2.9. Множества Жюлиа

```python
import numpy as np
import matplotlib.pyplot as plt

def julia(c=None, func=None, title=""):
    w, h = 800, 800
    julia = np.zeros((w, h))
    eps = 1e-12
    for ix in range(w):
        for iy in range(h):
            z = complex(ix / w * 4 - 2, iy / h * 4 - 2)
            nit = 0
            while abs(z) <= 10 and nit < 1000:
                if func == "custom1":
                    denom = 1 - z * z
                    if abs(denom) < eps:
                        break
                    z = (z * z + 1j) / denom
                elif func == "iz5":
                    denom = z ** 5 - 1j
                    if abs(denom) < eps:
                        break
                    z = 1j * (z ** 5 - 1) / denom
                elif func == "z2i":
                    denom = 1 - z * z
                    if abs(denom) < eps:
                        break
                    z = (z * z + 1j) / denom
                else:
                    z = z * z + c
                nit += 1
            julia[ix, iy] = nit / 1000
    plt.figure(figsize=(8, 8))
    plt.imshow(julia.T, cmap="hot", extent=[-2, 2, -2, 2])
    plt.title(title)
    plt.axis("off")
    plt.show()

julia(complex(-0.1, 0.65), title="c = -0.1 + 0.65i")
julia(complex(0.28, 0.009), title="c = 0.28 + 0.009i")
julia(func="custom1", title="f(z) = (z²+1)/(1−z²)")
julia(func="iz5", title="f(z) = i(z^5−1)/(z^5−i)")
julia(func="z2i", title="f(z) = (z²+i)/(1−z²)")
```

## 2.10. Стереографическая проекция на сферу Римана

```python
from mpl_toolkits.mplot3d import Axes3D

def stereo(x, y):
    d = 1 + x*x + y*y
    return 2*x/d, 2*y/d, (x*x + y*y - 1)/d

fig = plt.figure(figsize=(20,12))

t = np.linspace(-10,10,500)
X,Y,Z = stereo(t, np.zeros_like(t))
ax = fig.add_subplot(241, projection='3d')
ax.plot(X,Y,Z,'r',lw=2)
ax.set_title("Прямая")

theta = np.linspace(0,2*np.pi,500)
X,Y,Z = stereo(np.cos(theta), np.sin(theta))
ax = fig.add_subplot(242, projection='3d')
ax.plot(X,Y,Z,'r',lw=2)
ax.set_title("Окружность")

X,Y,Z = stereo(t, t*t)
ax = fig.add_subplot(243, projection='3d')
ax.plot(X,Y,Z,'r',lw=2)
ax.set_title("Парабола")

X,Y,Z = stereo(np.sinh(t), np.cosh(t))
ax = fig.add_subplot(244, projection='3d')
ax.plot(X,Y,Z,'r',lw=2)
ax.set_title("Гипербола")

phi = np.linspace(0, 50, 1000)
r = np.exp(0.1*phi)
X,Y,Z = stereo(r*np.cos(phi), r*np.sin(phi))
ax = fig.add_subplot(212, projection='3d')
ax.plot(X,Y,Z,'r',lw=2)
ax.set_title("Логарифмическая спираль r = e^{0.1φ}")

plt.tight_layout()
plt.show()
```
